```python
catalog = "alpina_dev"
```

```python
for s in ["landing", "bronze", "silver", "gold"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{s}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.landing.files")
```

```python
import shutil, os

import os
repo = os.getcwd()
print(repo)
print(os.listdir(repo))

vol  = f"/Volumes/{catalog}/landing/files"

paths = ["reference", "landing/members",
         "landing/claims/2026-10-01", "landing/claims/2026-10-02"]

for p in paths:
    shutil.copytree(f"{repo}/data/{p}", f"{vol}/{p}", dirs_exist_ok=True)

for p in paths:
    print(p, sorted(os.listdir(f"{vol}/{p}")))
```
